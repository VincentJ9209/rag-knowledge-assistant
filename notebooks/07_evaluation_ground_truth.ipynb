# Evaluation Ground Truth

This thin notebook inspects the frozen FAQ snapshot and validated ground-truth artifacts. Generation is implemented in `scripts/generate_ground_truth.py` so it is reusable and testable.

```python
from pathlib import Path
import json

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
evaluation_dir = project_root / 'data' / 'evaluation'
corpus_path = evaluation_dir / 'faq_llm_zoomcamp_139.json'
metadata_path = evaluation_dir / 'evaluation_metadata.json'

corpus = json.loads(corpus_path.read_text(encoding='utf-8'))
metadata = json.loads(metadata_path.read_text(encoding='utf-8'))
print('Frozen documents:', len(corpus))
print('Corpus SHA-256:', metadata['corpus_sha256'])
print('Snapshot date:', metadata['snapshot_date'])
```

```python
import csv
from collections import Counter

ground_truth_path = evaluation_dir / 'ground_truth.csv'
audit_path = evaluation_dir / 'ground_truth_audit_sample.csv'

if ground_truth_path.exists():
    with ground_truth_path.open(newline='', encoding='utf-8') as source:
        ground_truth = list(csv.DictReader(source))
    print('Ground-truth rows:', len(ground_truth))
    print('Split counts:', dict(Counter(row['split'] for row in ground_truth)))
    print('Human-audit sample:', audit_path)
else:
    print('Ground truth is not generated. Run:')
    print('uv run python -m scripts.generate_ground_truth --corpus data/evaluation/faq_llm_zoomcamp_139.json --output data/evaluation/ground_truth.csv --model gpt-5.4-mini --workers 4')
```
